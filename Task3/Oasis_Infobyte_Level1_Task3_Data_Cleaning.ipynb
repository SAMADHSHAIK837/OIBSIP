## Missing Data Handling

Missing values are handled according to the type and meaning of each column.

- `Customer_Age`: Median imputation is used because age is a numerical variable and the median is less affected by extreme values.
- `Customer_Rating`: Median imputation is used because ratings are numerical and the median provides a robust central value.
- Categorical columns such as `Gender`, `Region`, `Product_Name`, `Product_Category`, and `Payment_Method`: Mode imputation is appropriate because these columns contain categories.
- `Order_Date`: Missing dates will be investigated before deciding whether to fill or remove them because dates should not be replaced with an arbitrary numerical average.

```python
missing_report = pd.DataFrame({
    'Missing_Count': df.isnull().sum(),
    'Missing_Percentage': (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_report = missing_report[
    missing_report['Missing_Count'] > 0
]

missing_report
```

```python
# Numerical columns: median imputation
numeric_missing = [
    'Customer_Age',
    'Customer_Rating'
]

for col in numeric_missing:
    df[col] = df[col].fillna(df[col].median())


# Categorical columns: mode imputation
categorical_missing = [
    'Gender',
    'Region',
    'Product_Name',
    'Product_Category',
    'Payment_Method'
]

for col in categorical_missing:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
```

## Outlier Detection Using the IQR Method

Outliers are observations that are unusually high or low compared with the majority of the data.

The Interquartile Range (IQR) method is used to identify potential outliers in numerical columns.

The IQR is calculated as:

IQR = Q3 - Q1

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these bounds are considered potential outliers.


```python
numeric_columns = [
    'Customer_Age',
    'Quantity',
    'Unit_Price',
    'Discount',
    'Sales_Amount',
    'Customer_Rating'
]

outlier_report = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_report.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower_Bound': lower_bound,
        'Upper_Bound': upper_bound,
        'Outlier_Count': len(outliers)
    })

outlier_report = pd.DataFrame(outlier_report)

outlier_report
```

## Outlier Treatment Decision

The IQR method identified potential outliers in several numerical variables.

These observations will be retained rather than automatically removed. In a retail sales dataset, unusually high quantities, prices, or sales amounts can represent legitimate business transactions rather than data errors.

Removing valid transactions could lead to loss of useful information and inaccurate business analysis.

Therefore, the identified statistical outliers are retained unless they are confirmed to be data-entry errors.
